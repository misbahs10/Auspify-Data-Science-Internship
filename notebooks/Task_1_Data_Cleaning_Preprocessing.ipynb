## 1. Import the Dataset

## 2. Identify Missing Values

Missing values can affect data analysis and machine learning models. Therefore, we identify the number of missing values in each column before deciding how to handle them.

```python
missing_values = df.isnull().sum()

missing_values
```

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_summary = pd.DataFrame({
    "Missing Values": missing_values,
    "Missing Percentage": missing_percentage
})

missing_summary
```

### Duplicate Records

Duplicate records can lead to biased analysis and incorrect statistics. We check for duplicate rows before creating the cleaned dataset.

```python
duplicate_count = df.duplicated().sum()

print(f"Number of duplicate rows: {duplicate_count}")
```

```python
df[df.duplicated()]
```

### Duplicate Treatment

No duplicate records were found in the dataset, so no duplicate rows were removed.

```python
text_columns = [
    "type",
    "title",
    "director",
    "country",
    "rating",
    "duration",
    "listed_in"
]
```

```python
for column in text_columns:
    df[column] = df[column].astype("string").str.strip()
```

```python
df["director"].value_counts().head(10)
```

```python
(df == "Not Given").sum()
```

```python
df = df.replace("Not Given", pd.NA)
```

```python
df.isnull().sum()
```

### Standardizing Missing Values

Some fields use the text value `"Not Given"` to represent unavailable information. These placeholder values are converted to proper missing values (`NaN`) to maintain consistency during analysis.

```python
df["date_added"] = pd.to_datetime(
    df["date_added"],
    errors="coerce"
)
```

```python
df["date_added"].dtype
```

```python
print("Invalid dates:", df["date_added"].isna().sum())
```

```python
df["added_year"] = df["date_added"].dt.year
df["added_month"] = df["date_added"].dt.month
```

```python
df[["date_added", "added_year", "added_month"]].head()
```

```python
df["duration"].value_counts().head(20)
```

```python
df["duration_value"] = (
    df["duration"]
    .str.extract(r"(\d+)")
    .astype("Int64")
)
```

```python
df[["type", "duration", "duration_value"]].head(10)
```

```python
df["duration_unit"] = df["duration"].str.extract(r"([A-Za-z]+)")
```

```python
df[["duration", "duration_value", "duration_unit"]].head(10)
```

```python
categorical_columns = [
    "type",
    "rating",
    "duration_unit"
]
```

```python
for column in categorical_columns:
    print(f"\n{column}:")
    print(df[column].value_counts().head(10))
```

```python
print("Final Shape:", df.shape)
```

```python
df.isnull().sum()
```

```python
print("Duplicates:", df.duplicated().sum())
```

```python
df.dtypes
```

```python
df.head()
```

```python
output_path = "../data/netflix_cleaned.csv"

df.to_csv(output_path, index=False)

print(f"Cleaned dataset saved to: {output_path}")
```
